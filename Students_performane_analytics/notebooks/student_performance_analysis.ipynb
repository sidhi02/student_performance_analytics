```python
df = df.drop_duplicates()
```

```python
df.columns = df.columns.str.lower().str.replace(" ","_").str.replace("/","_")
df.head()
```

```python
df["total_score"] = df["math_score"] + df["reading_score"] + df["writing_score"]
```

```python
df ["average_score"] = df["total_score"] / 3
```

```python
df["result"] = np.where(
    (df["math_score"] >=35) &
    (df["reading_score"] >=35) &
    (df["writing_score"] >=35),
    "Pass",
    "Fail"
)
```

```python
def grade_category(avg):
    if avg >= 90:
        return "Excellent"
    elif avg >= 75:
        return "Very Good"
    elif avg >= 60:
        return "good"
    elif avg >= 35:
        return "Average"
    else:
        return "Fail"
df["grade_category"] = df["average_score"].apply(grade_category)
```

```python
df.head()
```

```python
avg_math = df["math_score"].mean()
avg_reading = df["reading_score"].mean()
avg_writing = df["writing_score"].mean()

print("Average Math Score:", avg_math)
print("Average Reading Score:", avg_reading)
print("Average Writing Score:", avg_writing)
```

```python
print("Highest Math Score:", df["math_score"].max())
print("Lowest Math Score:", df["math_score"].min())

print("Highest Reading Score:", df["reading_score"].max())
print("Lowest Reading Score:", df["reading_score"].min())

print("Highest Writing Score:", df["writing_score"].max())
print("Lowest Writing Score:", df["writing_score"].min())
```
